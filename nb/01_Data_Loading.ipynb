```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

```python
forest_file = "../data/Bird_Monitoring_Data_FOREST.xlsx"

forest_excel = pd.ExcelFile(forest_file)

forest_sheet_names = forest_excel.sheet_names

print("Forest Sheets:")
print(forest_sheet_names)
print(f"\nTotal Sheets: {len(forest_sheet_names)}")
```

```python
forest_sheets_dict = {
    sheet: forest_excel.parse(sheet)
    for sheet in forest_sheet_names
}

forest_df = pd.concat(
    [
        df.assign(Admin_Sheet=sheet_name)
        for sheet_name, df in forest_sheets_dict.items()
    ],
    ignore_index=True
)

forest_df.head()
```

```python
print("Forest Shape:")
print(forest_df.shape)
```

```python
grassland_file = "../data/Bird_Monitoring_Data_GRASSLAND.xlsx"

grassland_excel = pd.ExcelFile(grassland_file)

grassland_sheet_names = grassland_excel.sheet_names

print("Grassland Sheets:")
print(grassland_sheet_names)
print(f"\nTotal Sheets: {len(grassland_sheet_names)}")
```

```python
grassland_sheets_dict = {
    sheet: grassland_excel.parse(sheet)
    for sheet in grassland_sheet_names
}

grassland_df = pd.concat(
    [
        df.assign(Admin_Sheet=sheet_name)
        for sheet_name, df in grassland_sheets_dict.items()
    ],
    ignore_index=True
)

grassland_df.head()
```

```python
print("Grassland Shape:")
print(grassland_df.shape)
```

```python
print("="*50)
print("FOREST DATASET")
print("="*50)

print("Shape :", forest_df.shape)

print("\nColumns:")
print(forest_df.columns.tolist())

print("\nData Types:")
print(forest_df.dtypes)
```

```python
print("="*50)
print("GRASSLAND DATASET")
print("="*50)

print("Shape :", grassland_df.shape)

print("\nColumns:")
print(grassland_df.columns.tolist())

print("\nData Types:")
print(grassland_df.dtypes)
```

```python
forest_df.to_csv(
    "../cleaned_data/forest_raw.csv",
    index=False
)

grassland_df.to_csv(
    "../cleaned_data/grassland_raw.csv",
    index=False
)

print("Files Saved Successfully")
```

```python
import pandas as pd
import sqlite3

pd.set_option("display.max_columns", None)
```

```python
birds_df = pd.read_csv(
    "../cleaned_data/birds_cleaned.csv"
)

birds_df.head()
```

```python
conn = sqlite3.connect(
    "../database/bird_species.db"
)

print("Database Connected")
```

```python
birds_df.to_sql(
    "bird_observations",
    conn,
    if_exists="replace",
    index=False
)

print("Table Created Successfully")
```

```python
query = """
SELECT *
FROM bird_observations
LIMIT 5
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT COUNT(*) AS Total_Observations
FROM bird_observations
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT COUNT(DISTINCT Scientific_Name)
AS Total_Species
FROM bird_observations
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    Common_Name,
    COUNT(*) AS Observations
FROM bird_observations
GROUP BY Common_Name
ORDER BY Observations DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    Habitat,
    COUNT(*) AS Observations
FROM bird_observations
GROUP BY Habitat
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    Admin_Unit_Code,
    COUNT(*) AS Observations
FROM bird_observations
GROUP BY Admin_Unit_Code
ORDER BY Observations DESC
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    Habitat,
    COUNT(DISTINCT Scientific_Name)
    AS Unique_Species
FROM bird_observations
GROUP BY Habitat
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT
    Flyover_Observed,
    COUNT(*) AS Count
FROM bird_observations
GROUP BY Flyover_Observed
"""

pd.read_sql(query, conn)
```
